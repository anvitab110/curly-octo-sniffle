# Quick Start 

```python
from Bio.Seq import Seq
from Bio import SeqIO
```

```python
#sequences
seq = Seq("AGTACATGACCTTG") #creates a sequence object (5 to 3)
print(seq)

print(seq.complement()) #the complementary dna strand (3 to 5)
print(seq.reverse_complement()) #the complementary dna strand reversed (5 to 3)
```

```python
for seq_record in SeqIO.parse("ls_orchid.fasta", "fasta"): #given a fasta file on local 
    print(seq_record.id) #id associated with that sequence
    print(repr(seq_record.seq)) #the actual sequence
    print(len(seq_record)) #length of sequence
```

```python
coding_dna = Seq("ATGGCCATTGTAATGGGCCGCTGAAAGGGTGCCCGATAG") #rna product will resemble this
template_dna = coding_dna.reverse_complement() #rna product will be created to match this (in the opposite)
template_dna
```

```python
messenger_rna = coding_dna.transcribe()
messenger_rna
```

```python
#OR 
messenger_rna = template_dna.reverse_complement().transcribe()
messenger_rna

#back_transribe() will reverse the mrna into dna 
```

```python
messenger_rna.translate(stop_symbol = '!') #gives the protein sequence! 
#does not need to be done using the rna only, can be done using coding dna 
```

```python
#asterisk/stop_symbol argument in protein sequence means stop codon 
#specifying a translation table name/number from NCBI will fix the stopping 
print(coding_dna.translate(table = "Vertebrate Mitochondrial"))
print(coding_dna.translate(table = 2))
```

```python
#adding to_stop argument means translation finishes at the first stop codon seen 
coding_dna.translate(table=2, to_stop=True)
```

```python
from Bio.SeqFeature import SeqFeature, SimpleLocation
```

```python
#seqfeature object does not contain any specific sequence
seq = Seq("ACCGAGACGGCAAAGGCTAGCATAGGTATGAGACTTCCTTCCTGCCAGTGCTGAGGAACTGGGAGCCTAC")
feature = SeqFeature(SimpleLocation(5, 18, strand = -1), type = "gene")
feature_seq = seq[feature.location.start : feature.location.end].reverse_complement()
print(feature_seq)
#seqfeature object describes how to get the location of interest 
```

```python
feature_seq = feature.extract(seq) #simpler way of doing the above, inherently does the joining etc on a specified sequence
print(feature_seq)
```
